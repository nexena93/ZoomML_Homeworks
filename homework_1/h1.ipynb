### Question 6: median value of horsepower
1. Find the median value of the horsepower column in the dataset:

```python
df.columns
```

```python
horsepower = df["horsepower"]
med_horse = horsepower.median()
print("Median value of the horsepower column: ", med_horse)
```

2. Next, calculate the most frequent value of the same horsepower column:

```python
grouped = df.groupby("horsepower").size()
grouped.reset_index(name = "count")
```

```python
print(grouped.idxmax(), grouped.max())
```

```python
most_frequent_horse = grouped.idxmax()
print("Most frequent value of horsepower column: ", most_frequent_horse)
```

3. Use the 'fillna' method to fill the missing values in the horsepower column with the most frequent value from the previous step:

```python
df["new_horsepower"] = df["horsepower"].fillna(most_frequent_horse)
print(df["new_horsepower"].hasnans)
print(df["horsepower"].hasnans)
```

4. Now, calculate the median value of horsepower once again:

```python
new_horsepower = df["new_horsepower"]
new_med_horse = new_horsepower.median()
```

```python
print("Old median horsepower: ", med_horse)
print("New median horsepower: ", new_med_horse)
print("Difference between the two values: ", med_horse-new_med_horse)
```

7. Get the underlying NumPy array. Let's call it X.

```python
import numpy as np
```

```python
X = (
    df
    [asian_cars]
    [["vehicle_weight","model_year"]]
    .head(7)
    .to_numpy() 
    )
X
```

5. Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.

```python
XTX = X.T@X
XTX
```

6. Invert XTX.

```python
m = np.linalg.inv(XTX)
m
```

7. Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

8. Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.

```python
w = (m@X.T)@ y
w
```

9. What's the sum of all the elements of the result?

```python
print("Sum of the element of w: ", w.sum())
```
